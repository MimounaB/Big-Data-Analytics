```python
import time

data = []

for page in range(1, 495):
    url = f"https://arabic-quotes.com/quotes?page={page}"
    response = requests.get(url)

    soup = BeautifulSoup(response.text, "html.parser")
    quotes = soup.find_all("p", class_="quote-text")

    for quote in quotes:
        container = quote.parent.parent
        author = container.find("p", class_="quote-author")

        quote_text = quote.get_text(strip=True)
        author_text = author.get_text(strip=True)

        data.append({
            "quote": quote_text,
            "author": author_text
        })

    time.sleep(0.3)

print(len(data))
```

```python
print(len(data))
print(data[:3])
```

```python
unique_quotes = set(item["quote"] for item in data)
print(len(unique_quotes))
```

```python
from collections import Counter

quote_counts = Counter(item["quote"] for item in data)

print(quote_counts.most_common(5))
```

```python
unique_data = list(set((item["quote"], item["author"]) for item in data))

print(len(unique_data))
```

```python
data_clean = [
    {"quote": quote, "author": author}
    for quote, author in unique_data
]
```

```python
import pandas as pd
df = pd.DataFrame(data_clean)
print(df.shape)
print(df.head())
```
